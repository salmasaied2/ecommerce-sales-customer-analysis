# 1. Import Libraries & Upload Dataset

```python
# Important required data analysis libraries
from google.colab import files
import pandas as pd
import numpy as np

# Upload the dataset directly from your local machine
uploaded = files.upload()

# Automatically retrieve the uploaded file name
file_name = list(uploaded.keys())[0]
print(f"File uploaded successfully: {file_name}")

# Read the uploaded CSV file into a DataFrame
df = pd.read_csv(file_name)
```

```python
# Display the first 5 rows to verify the data structure
df.head()
```

# 2. Data Inspection & Overview

```python
# Check dataset dimensions (rows, columns)
print(f"Dataset Shape: {df.shape}")

# Inspect data types and missing values counts
df.info()

# Check for any duplicated rows in the dataset
print(f"Number of duplicate rows: {df.duplicated().sum()}")

# Generate descriptive statistics for numerical columns
df.describe()
```

# 3. Data Cleaning & Feature Engineering

```python
# 1. Handle missing values for review_score and gender
# Fill missing review scores with the median value
df['review_score'] = df['review_score'].fillna(df['review_score'].median())

# Fill missing gender values with 'Unknown' to avoid grouping issues
df['gender'] = df['gender'].fillna('Unknown')

# 2. Convert 'order_date' to proper datetime format
df['order_date'] = pd.to_datetime(df['order_date'], errors='coerce')

# 4. Filter out any illogical negative/zero quantities or prices
df = df[(df['quantity'] > 0) & (df['price'] > 0)]

# 5. Create new calculated columns for business analysis
# Total Revenue per transaction
df['revenue'] = df['quantity'] * df['price']

# Extract Year-Month for temporal trend analysis
df['month_year'] = df['order_date'].dt.to_period('M')

# Verify that missing values are completely gone and new columns are added
print("Missing values check after cleaning:")
print(df[['review_score', 'gender', 'revenue']].isnull().sum())

# Display the first few rows of the clean dataset
df.head(20)
```

# 4. Exploratory Data Analysis(EDA)

```python
# 1. What are the top 5 best-selling products by total revenue?
top_products = df.groupby('product_name')['revenue'].sum().reset_index()
top_products = top_products.sort_values(by='revenue', ascending=False).head()
print("--- Top 5 Best-Selling Products by Revenue ---")
print(top_products, "\n")

# 2. What cities generate the most revenue?
top_cities = df.groupby('city')['revenue'].sum().reset_index()
top_cities = top_cities.sort_values(by='revenue', ascending=False).head(5)
print("--- Top 5 Cities by Revenue ---")
print(top_cities, "\n")

# 3. Overall Business KPIs (Total Revenue, Total Orders, Average Order Value)
total_revenue = df['revenue'].sum()
total_orders = df['customer_id'].count() # Total transactions
average_order_value = total_revenue / total_orders

print("--- Key Performance Indicators (KPIs) ---")
print(f"Total Revenue: ${total_revenue:,.2f}")
print(f"Total Transactions: {total_orders}")
print(f"Average Order Value (AOV): ${average_order_value:,.2f}")
```

# 5. Simple Customer Analysis (RFM)

```python
# 1. Reference date for Recency
snapshot_date = df['order_date'].max() + pd.Timedelta(days=1)

# 2. Calculate simple metrics per customer
rfm_df = df.groupby('customer_id').agg({
    'order_date': lambda x: (snapshot_date - x.max()).days, # Recency: Days since last order
    'customer_id': 'count',                                  # Frequency: Total orders count
    'revenue': 'sum'                                         # Monetary: Total spent money
}).rename(columns={
    'order_date': 'recency',
    'customer_id': 'frequency',
    'revenue': 'monetary'
}).reset_index()

# 3. Simple Segmentation based on Monetary (Total Spend)
# Customers who spent more than the average are 'VIP', others are 'Regular'
avg_spend = rfm_df['monetary'].mean()

def simple_segment(monetary_value):
    if monetary_value >= avg_spend:
        return 'VIP Customers'
    else:
        return 'Regular Customers'

rfm_df['customer_segment'] = rfm_df['monetary'].apply(simple_segment)

# Display the summary of customer segments
print("--- Customer Segments Count ---")
print(rfm_df['customer_segment'].value_counts())

# Display the first few rows
print("\n--- Sample of Customer Analysis ---")
print(rfm_df.head())
```

```python
# 1. Save the main cleaned transactions dataset to CSV
df.to_csv('cleaned_online_retail_data.csv', index=False)

# 2. Save the customer segmentation (RFM) dataset to CSV
rfm_df.to_csv('customer_rfm_segments.csv', index=False)

print("Files saved successfully! You can now download them from the left panel in Colab.")

# Optional: Automatically download the files to your local machine
from google.colab import files
files.download('cleaned_online_retail_data.csv')
files.download('customer_rfm_segments.csv')
```
